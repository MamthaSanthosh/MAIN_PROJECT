# Project Title

HOSPITAL BILLING AND TREATMENT ANALYSIS

# **Objective(Problem statment)**

1. To analyze patient billing data and understand overall hospital expenses.

2. To identify the departments, treatments, and room types associated with higher costs.

3. To study how hospital stay duration, medicine costs, and test costs affect the total bill.

4. To compare different admission types and their impact on patient billing.

5. Identifying missing values,duplicate values,inconsistents.

5. To generate meaningful insights that support better hospital financial and operational decisions.


# Outcome

1. The dataset contains 1,000 patient billing records and 10 columns.

2. It includes information about patients, departments, treatments, admission types, and room types.

3. It also contains stay duration, medicine cost, test cost, and total bill.

4. The dataset can be used to analyze treatment expenses and factors affecting hospital billing.

# Domain
Healthcare/Hospital Management

#**Dataset Information**

**Source:**  WHO Portals

**Dataset Column/features Description:**
**1000 Rows and 10 Columns**

* **Bill_ID**- Unique identification number for each billing record.

* **Patient_ID** - Unique identification number assigned to each patient.

* **Department** - Hospital department where the patient received treatment.

* **Treatment** - Type of medical treatment provided to the patient.

* **Admission_Type** - Indicates how the patient was admitted, such as Emergency, Urgent, or Elective.

* **Room_Type** - Type/category of room occupied by the patient during the hospital stay.

* **Stay_Days** - Number of days the patient stayed in the hospital.

* **Medicine_Cost** - Amount spent on medicines during the patient's treatment.

* **Test_Cost** - Amount spent on medical/laboratory tests.

* **Total_Bill** - Total amount billed to the patient for the hospital stay and treatment.

*GROUPED ANALYSIS*

```python
df.groupby('Department').size()     #finding size of department
```

```python
department_count = df.groupby('Department').size().reset_index(name='Patient_Count')     #finding size /count of department and alias as patient count

print(department_count)
```

```python
df.groupby('Department')['Total_Bill'].mean()    #finding mean value for the department wise total bill
```

```python
department_bill = df.groupby('Department')['Total_Bill'].mean().reset_index()   #finding mean value for total bill by department wise

print(department_bill)
```

```python
department_total = df.groupby('Department')['Total_Bill'].sum().reset_index()   #finding sum of total bill by department

print(department_total)
```

```python
grouped_analysis = df.groupby('Department').agg(                 #finding aggregate value for columns
    Patient_Count=('Patient_ID', 'count'),
    Average_Room_Charges=('Room_Charges', 'mean'),
    Average_Treatment_Charges=('Treatment_Charges', 'mean'),
    Total_Revenue=('Total_Bill', 'sum'),
    Average_Bill=('Total_Bill', 'mean')
).reset_index()
print(grouped_analysis)
```

**INTERPRETATION**

*The grouped analysis provides a department-wise view of patient volume, room charges, treatment charges, and billing. It helps hospital management identify departments with higher patient volumes and higher financial contribution. Comparing average room and treatment charges can highlight differences in the cost structure between departments, while total revenue and average bill provide useful information for budget planning, resource allocation, cost monitoring, and departmental performance analysis.*
